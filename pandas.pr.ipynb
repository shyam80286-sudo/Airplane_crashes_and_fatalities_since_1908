```python
# Load the dataset


# 1. Fill missing categorical and text values
categorical_cols = [
    "Location", "Operator", "Flight #", "Route",
    "AC Type", "Registration", "cn/ln",
    "Summary", "Summary_Clean", "Extracted_Keyword"
]

for col in categorical_cols:
    if col in df_csv.columns:
        df_csv[col] = df_csv[col].fillna("Unknown")

# 2. Fill missing Time values
if "Time" in df_csv.columns:
    df_csv["Time"] = df_csv["Time"].fillna("Unknown")

# 3. Fill missing numeric values using the median
numeric_cols = [
    "Aboard", "Aboard Passangers", "Aboard Crew",
    "Fatalities", "Fatalities Passangers",
    "Fatalities Crew", "Ground"
]

for col in numeric_cols:
    if col in df_csv.columns:
        df_csv[col] = pd.to_numeric(
            df_csv[col], errors="coerce"
        )
        df_csv[col] = df_csv[col].fillna(
            df_csv[col].median()
        )

# 4. Recalculate derived rates where possible
if "Aboard" in df_csv.columns and "Fatalities" in df_csv.columns:
    valid = df_csv["Aboard"] > 0

    df_csv["Fatality_Rate"] = (
        df_csv["Fatalities"] / df_csv["Aboard"]
    ).where(valid)

    df_csv["Survival_Rate"] = (
        (df_csv["Aboard"] - df_csv["Fatalities"])
        / df_csv["Aboard"]
    ).where(valid)

    df_csv["Fatality_Percentage"] = (
        df_csv["Fatality_Rate"] * 100
    )

    df_csv["Survival_Percentage"] = (
        df_csv["Survival_Rate"] * 100
    )

# 5. Check remaining missing values
print("Missing values after cleaning:")
print(df_csv.isnull().sum()[df_csv.isnull().sum() > 0])

# 6. Display the first five rows
print(df_csv.head())


# 7. Save the cleaned dataset


print("Cleaned dataset saved successfully!")
```

```python

search_terms = [
    "weather",
    "engine failure",
    "mechanical failure",
    "pilot error",
    "fire",
    "collision",
    "takeoff",
    "landing",
    "storm",
    "fog"
]

summary_text = df_csv["Summary_Clean"]

term_results = []

for term in search_terms:
    matching_rows = summary_text.str.contains(
        term, case=False, regex=False, na=False
    )

    term_results.append({
        "Search_Term": term,
        "Matching_Records": int(matching_rows.sum())
    })

term_analysis = (
    pd.DataFrame(term_results)
    .sort_values("Matching_Records", ascending=False)
    .reset_index(drop=True)
)

display(term_analysis)
```

```python

summary_text = df_csv["Summary_Clean"]

patterns = {
    "Weather-related terms": r"weather|storm|fog|wind|icing|rain",
    "Mechanical-related terms": r"engine failure|mechanical failure|engine malfunction|technical failure",
    "Fire-related terms": r"\bfire\b|burning|flames",
    "Collision-related terms": r"collision|collided|mid-air collision",
    "Takeoff-related terms": r"takeoff|take-off",
    "Landing-related terms": r"landing|landed|forced landing"
}

category_results = []

for category, pattern in patterns.items():
    matches = summary_text.str.contains(
        pattern, case=False, regex=True, na=False
    )

    category_results.append({
        "Category": category,
        "Matching_Records": int(matches.sum()),
        "Percentage_of_All_Records": round(
            matches.sum() / len(df_csv) * 100, 2
        ) if len(df_csv) else 0
    })

category_analysis = (
    pd.DataFrame(category_results)
    .sort_values("Matching_Records", ascending=False)
    .reset_index(drop=True)
)

display(category_analysis)
```

```python

# Calculate accident counts for each year
yearly_window = (
    df_csv.dropna(subset=["Year"])
    .groupby("Year")
    .size()
    .sort_index()
    .rename("Accident_Count")
    .to_frame()
)

display(yearly_window.head(10))
```

```python

# Previous recorded year's accident count
yearly_window["Previous_Year_Count"] = (
    yearly_window["Accident_Count"].shift(1)
)

# Absolute change in recorded accident counts
yearly_window["Change_From_Previous_Year"] = (
    yearly_window["Accident_Count"]
    - yearly_window["Previous_Year_Count"]
)

display(yearly_window.tail(15))
```

```python

yearly_window["Percentage_Change"] = (
    yearly_window["Accident_Count"]
    .pct_change(fill_method=None) * 100
).round(2)

display(
    yearly_window[
        [
            "Accident_Count",
            "Previous_Year_Count",
            "Change_From_Previous_Year",
            "Percentage_Change"
        ]
    ].tail(15)
)
```

```python

yearly_window["Three_Year_Rolling_Average"] = (
    yearly_window["Accident_Count"]
    .rolling(window=3, min_periods=3)
    .mean()
    .round(2)
)

display(yearly_window.tail(15))
```

```python


plt.figure(figsize=(14, 6))

plt.plot(
    yearly_window.index,
    yearly_window["Accident_Count"],
    label="Yearly Accident Count"
)

plt.plot(
    yearly_window.index,
    yearly_window["Three_Year_Rolling_Average"],
    label="Three-Year Rolling Average"
)

plt.title("Yearly Aviation Accidents and Rolling Average")
plt.xlabel("Year")
plt.ylabel("Recorded Accident Count")
plt.legend()
plt.tight_layout()
plt.show()

print("Early years: From the beginning of the 1900s to the 1920s, the number of recorded accidents was relatively low.")
print("Increasing trend: During the 1930s and early 1940s, accident counts increased considerably.")
print("Peak period: The highest annual accident count occurred around the mid-1940s, reaching approximately 88 accidents.")
print("Mid-century fluctuations: From the 1950s through the 1990s, accident counts fluctuated considerably, generally remaining at higher levels than in the early decades.")
print("Three-year rolling average: The orange line smooths short-term fluctuations by averaging accident counts over three consecutive years. This makes the overall trend easier to identify.")
print("Recent decline: From the early 2000s onward, the graph shows a substantial decline in recorded accidents, reaching fewer than 10 in the final year shown.")
```

```python
keyword_patterns = {
    "Weather-related terms": r"\b(weather|storm|fog|rain|snow|ice|icing|wind)\b",
    "Engine-related terms": r"\b(engine|engines|engine failure)\b",
    "Mechanical-related terms": r"\b(mechanical|malfunction|failure|technical)\b",
    "Pilot-related terms": r"\b(pilot|crew error|human error)\b",
    "Fire-related terms": r"\b(fire|smoke|explosion)\b",
    "Takeoff or landing terms": r"\b(takeoff|take-off|landing|approach)\b"
}

summary_text = df_csv["Summary"].fillna("").str.lower()

keyword_results = []

for category, pattern in keyword_patterns.items():
    count = summary_text.str.contains(
        pattern,case=False,regex=True, na=False
    ).sum()

    keyword_results.append({
        "Keyword_Category": category,
        "Records_Mentioning_Terms": int(count),
        "Percentage_of_All_Records": round(
            count / len(df_csv) * 100, 2
        ) if len(df_csv) else 0
    })

keyword_summary = pd.DataFrame(keyword_results).sort_values(
    "Records_Mentioning_Terms", ascending=False
)

display(keyword_summary)
```
